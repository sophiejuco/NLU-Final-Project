## Submitting GPU Job:


Getting GPU Instance
```
srun --account=ds_ga_1011-2024fa --partition=PARTITION_ID  --gres=gpu:1 --time=4:00:00 --pty /bin/bash
```

Available GPU Partitions
- `n1s8-v100-1`: 1 V100 GPU
- `g2-standard-12`: 1 L4 GPU,  12 vCPUs, 48GB memory
- `g2-standard-24`: 2 L4 GPUs, 24 vCPUs, 96GB memory
- `g2-standard-48`: 4 L4 GPUs, 48 vCPUs, 192GB memory
- `c12m85-a100-1`: 1 A100 40GB GPU


Getting your file into the correct directory:

Navigate to `scratch`

```
cd /scratch/sgr344
```

If your code is not in `scratch` (as checked by the command `ls`), copy your file into the instance:

```
scp -r greene-dtn:/scratch/sgr344/FILE_PATH .
```
Note: the `-r` tag is used for copying in directories.

If you encounter a "WARNING: REMOTE HOST HAS CHANGED" error, run:

```
ssh-keygen -R greene-dtn
```
And then re-run previous `scp` command.
